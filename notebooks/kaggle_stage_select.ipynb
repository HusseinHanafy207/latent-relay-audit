# Evidence-preserving selection diagnostic on Kaggle T4

Same Tesla T4 as Stages A–C. Do not switch GPUs.

Uses only `data/dev.jsonl` (20 examples). Do not load the held-out split.

Question: can deliberately keeping the evidence line repair the compressed relay at budget 32?

One sender cache per example, reused across five receiver conditions:
1. Full relay (reference)
2. Ordinary headwise attention
3. Random selection at the same budget
4. Force the complete relevant inventory line, then fill remaining slots with the original attention rule
5. Control: reserve the same number of slots for an unrelated inventory line

Sink tokens and latent states stay identical. Forced tokens occupy part of the existing 32-slot budget; they are not extra information. OBF is not used.

This is a diagnostic intervention, not a proposed compressor. 100 receiver evaluations. Resume is on. Download `/kaggle/working/results_stage_select` during the run.

```python
import os, json, shutil, sys
from pathlib import Path

os.environ["LATENT_RELAY_INFER_DTYPE"] = "fp16"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

ON_KAGGLE = Path("/kaggle/working").exists()
WORK = Path("/kaggle/working") if ON_KAGGLE else Path.cwd()

candidates = [Path.cwd()]
if Path("/kaggle/input").is_dir():
    for root, dirs, files in os.walk("/kaggle/input"):
        dirs[:] = [d for d in dirs if d not in {".git", ".pytest_cache"}]
        if "stage_select.py" in files and Path(root).name == "src":
            candidates.append(Path(root).parent)

REPO = None
for cand in candidates:
    if (cand / "src" / "stage_select.py").is_file() and (cand / "data" / "dev.jsonl").is_file():
        REPO = cand
        break
if REPO is None:
    raise FileNotFoundError("Need src/stage_select.py and data/dev.jsonl. Re-upload the repo zip.")

if ON_KAGGLE:
    dest = WORK / "latent-relay-audit"
    if str(REPO).startswith("/kaggle/input"):
        shutil.copytree(REPO, dest, dirs_exist_ok=True)
    elif not (dest / "src" / "stage_select.py").is_file():
        shutil.copytree(REPO, dest, dirs_exist_ok=True)
    REPO = dest
    os.chdir(REPO)
else:
    os.chdir(REPO)

if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
print("REPO", REPO.resolve())
print("dev examples", sum(1 for _ in (REPO / "data" / "dev.jsonl").open(encoding="utf-8")))
```
