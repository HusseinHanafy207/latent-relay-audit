# Question withheld from A on Kaggle T4

Same Tesla T4. Do not switch GPUs.

Uses only `data/validate.jsonl` (20 fresh examples). Do **not** load `data/test.jsonl`.

A encodes the inventory and never sees the eventual question or choices. The question is then revealed to the selector and to B.

Five receiver conditions, one sender cache per example:
1. Full relay
2. Recency: last 32 prompt positions (now inventory tokens, not A's question)
3. Correct-question probe
4. Generic probe (no package name)
5. Question-name record matching: parse the package from B's question and keep that inventory line. No gold locker or record index.

Shared answer extractor for every method. Original first-match scores are stored as `kind_legacy` / `correct_legacy`.

Resume is on. If `results_stage_withhold/` from the previous 80 evaluations is in the upload, only the 20 match evaluations run. 100 receiver evaluations when starting from scratch. Budget 32, sender-only with filler. OBF is unused. Download `/kaggle/working/results_stage_withhold` during the run.

```python
import os, json, shutil, sys
from pathlib import Path

os.environ["LATENT_RELAY_INFER_DTYPE"] = "fp16"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

ON_KAGGLE = Path("/kaggle/working").exists()
WORK = Path("/kaggle/working") if ON_KAGGLE else Path.cwd()

candidates = [Path.cwd()]
if Path("/kaggle/input").is_dir():
    for root, dirs, files in os.walk("/kaggle/input"):
        dirs[:] = [d for d in dirs if d not in {".git", ".pytest_cache"}]
        if "stage_withhold.py" in files and Path(root).name == "src":
            candidates.append(Path(root).parent)

REPO = None
for cand in candidates:
    if (cand / "src" / "stage_withhold.py").is_file() and (cand / "data" / "validate.jsonl").is_file():
        REPO = cand
        break
if REPO is None:
    raise FileNotFoundError("Need src/stage_withhold.py and data/validate.jsonl. Re-upload the repo zip.")

if ON_KAGGLE:
    dest = WORK / "latent-relay-audit"
    if str(REPO).startswith("/kaggle/input"):
        shutil.copytree(REPO, dest, dirs_exist_ok=True)
    elif not (dest / "src" / "stage_withhold.py").is_file():
        shutil.copytree(REPO, dest, dirs_exist_ok=True)
    REPO = dest
    os.chdir(REPO)
else:
    os.chdir(REPO)

if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
print("REPO", REPO.resolve())
print("validate examples", sum(1 for _ in (REPO / "data" / "validate.jsonl").open(encoding="utf-8")))
```
