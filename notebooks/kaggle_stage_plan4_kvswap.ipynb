# Key/value diagnostic on frozen sep compact KV, plan4_dev, Kaggle T4

Same Tesla T4. Do not switch GPUs. Same Qwen3-4B revision, FP16, greedy decoding.

`plan4_dev` only. Do **not** load `plan4_hold` or `plan4_fresh`. Do not resume boundary, recon, iface, support, or hold.

A is not regenerated. Frozen compact IDs come from `results_stage_plan4_iface/a_artifacts.jsonl`.

Layout is frozen **sep**: first four positions, newline separator, result lines, consecutive positions, instruct framing. B's instructions, decoding, and scoring stay unchanged.

Four packed caches of the same token sequence:

1. **transfer** — A's keys and A's values.
2. **recon_k_a_v** — reconstructed keys with A's values.
3. **a_k_recon_v** — A's keys with reconstructed values.
4. **reconstruct** — reconstructed keys and values. This should recover the sep reconstruction reference.

This is a diagnostic. It requires recomputation. Do not train an adapter yet.

**Run cell 4 first** (`--phase check`). Identity mixes must recover the transfer and reconstruct caches. If `all_passed` is false, stop.

If the check passes, run cell 5 (`--phase compare`): four conditions × 90 cases (360 receiver evaluations).

Download `/kaggle/working/results_stage_plan4_kvswap`.

```python
import os, json, shutil, sys
from pathlib import Path

os.environ["LATENT_RELAY_INFER_DTYPE"] = "fp16"
os.environ["TOKENIZERS_PARALLELISM"] = "false"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

ON_KAGGLE = Path("/kaggle/working").exists()
WORK = Path("/kaggle/working") if ON_KAGGLE else Path.cwd()

candidates = [Path.cwd()]
if Path("/kaggle/input").is_dir():
    for root, dirs, files in os.walk("/kaggle/input"):
        dirs[:] = [d for d in dirs if d not in {".git", ".pytest_cache"}]
        if "stage_plan4_kvswap.py" in files and Path(root).name == "src":
            candidates.append(Path(root).parent)

REPO = None
for cand in candidates:
    if (cand / "src" / "stage_plan4_kvswap.py").is_file() and (cand / "data" / "plan4_dev.jsonl").is_file():
        REPO = cand
        break
if REPO is None:
    raise FileNotFoundError("Need src/stage_plan4_kvswap.py and data/plan4_dev.jsonl.")

if ON_KAGGLE:
    dest = WORK / "latent-relay-audit"
    if str(REPO).startswith("/kaggle/input"):
        shutil.copytree(REPO, dest, dirs_exist_ok=True)
    elif not (dest / "src" / "stage_plan4_kvswap.py").is_file():
        shutil.copytree(REPO, dest, dirs_exist_ok=True)
    REPO = dest
    os.chdir(REPO)
else:
    os.chdir(REPO)

if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
print("REPO", REPO.resolve())
print("plan4_dev", sum(1 for _ in (REPO / "data" / "plan4_dev.jsonl").open(encoding="utf-8")))
assert not (REPO / "data" / "plan4_hold.jsonl").samefile(REPO / "data" / "plan4_dev.jsonl")
art = REPO / "results_stage_plan4_iface" / "a_artifacts.jsonl"
print("frozen A artifacts", art.is_file(), art)
```
